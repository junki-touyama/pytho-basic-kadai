```python
class Human:

  def __init__(self, name, age):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print(f"{self.name}は、大人です。")
    else:
      print(f"{self.name}は、まだ子どもです。")

a = Human("當山純輝", 32)
b = Human("當山姫子", 32)
c = Human("當山もも", 5)
humans = [a, b, c]

a.check_adult()
b.check_adult()
c.check_adult()

for human_obj in humans:
      human_obj.check_adult()


```
